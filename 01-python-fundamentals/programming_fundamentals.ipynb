## The DRY Principle

The DRY principle is a fundamental philosophy that underpins efficient coding. It advocates for eliminating redundant code, encouraging you to consolidate repetitive instructions into a single, reusable block. This approach isn't merely about saving keystrokes; it's about reducing the surface area for errors, inconsistencies, and maintenance headaches.Let's say you're developing an e-commerce platform. Without DRY, you might find yourself copying and pasting code for tasks like calculating discounts, validating payment information, or sending order confirmation emails. This scattered repetition makes your codebase bloated and prone to errors. If a change is required, you'd have to hunt down and modify every instance of the duplicated code, increasing the risk of introducing bugs.DRY, on the other hand, encourages you to create a centralized function for each of these tasks. For example, a `calculate_discount(price, discount_percentage)` function would encapsulate the logic for calculating discounts, regardless of the specific product or promotion. This function acts as a single source of truth, ensuring consistency and simplifying updates.

## Understanding Code ExecutionBefore your code ever runs, Python parses and compiles the entire file into bytecode. This is why a syntax error on line 200 will stop your program before line 1 ever executes — Python already scanned the whole thing and choked on it during this compile pass.Once compiled to bytecode, execution proceeds sequentially, top to bottom, one statement at a time, with real, immediate effects as it goes. If line 5 throws a runtime error (not a syntax error — think dividing by zero, or calling an undefined variable), lines 1–4 already ran and did whatever they did. Line 6 never runs.

## Common Code Execution Pitfalls: How to Avoid Them

### 1. Indentation ErrorsA simple if-else statement where the else block is incorrectly indented:

````python
x = 0if x > 5:    print("x is greater than 5")  else:  # Incorrect indentation    print("x is less than or equal to 5")
````

````python
flag = Falseif flag:    x = 10    print(x) # No problem hereprint(x) # No error if flag = True
````

### 3. Immutable ValuesIn Python, strings are immutable, meaning their contents cannot be changed after they're created. The code below attempts to change the first character of the string "Hello" from 'H' to 'h'. However, this will result in a TypeError because strings are immutable.

````python
value = "Hello"value[0] = 'h'print(value)
````

### 4. Variable ReassignmentYou can accidentally use the same variable name again within your code. This can lead to unexpected results, as the later use of the name will change the variable's value. In this example, we first assign the value 5 to the variable x. Then, inside the if statement, we use x again and assign it the value 10. This changes the original value of x. So, when we print x after the if statement, it shows the new value, 10.

````python
x = 5if True:   x = 10  # You've used 'x' again here!   print(x)  # Output: 10print(x)  # Output: 10
````

### Reference Notes: Compile-Time vs. Runtime- **Compile time (whole file, before anything runs):** syntax problems — mismatched parens, bad indentation, invalid tokens. Also, scope itself (local vs. global/nonlocal) is *decided* at compile time based on whether a name is assigned anywhere in the function.- **Runtime (sequential, only when that line is reached):** whether a variable *actually has a value* when execution gets there, dividing by zero, calling a method that doesn't exist. This produces the classic `UnboundLocalError` gotcha — the compiler already decided `x` is local because it's assigned somewhere in the function, but referencing it *before* that assignment line executes blows up at runtime, not before.- **The reliable tell:** a `Traceback (most recent call last)` in the error output means it's a runtime error — a traceback is a call stack, and a call stack only exists because code was actually executing. A `SyntaxError` (including `IndentationError`/`TabError`, which are technically subclasses of `SyntaxError`) shows no traceback, just a file/line pointer, because execution never started.- **Checking compile-only, without running code:**  - `python3 -m py_compile your_script.py` — compiles and reports syntax errors without executing anything.  - `compile(source, "your_script.py", "exec")` — same idea, callable from within Python itself.  - Neither of these catches scope/name errors, undefined variables, or type mismatches — those only manifest when the code actually runs. Linters (pylint, flake8) go a step further with static analysis to *guess* at likely runtime problems without running the code, but they aren't the compiler.

A `class` is a blueprint for creating objects that bundle data (attributes) and behavior (methods) together. You define a class using the `class` keyword followed by the class name and a colon.

### ObjectsThey represent real-world entities or abstract concepts within your code. Each object is like a self-contained unit with its own data and the ability to perform actions.### ClassesThey serve as blueprints or templates for creating objects. Classes define the common characteristics (attributes) and behaviors (methods) that all objects of that class share.### EncapsulationKeeping the details of how an object works hidden inside the object, so you can change those details without breaking the rest of your program. It bundles data and the actions that operate on that data within a class, protecting the internal state of an object.### AbstractionSimplifies complex systems by focusing on essential features and interactions, hiding unnecessary details. It focuses on what an object does rather than how it does it, so your code is easier to understand and use.### InheritanceLets you create new kinds of objects based on existing ones, inheriting their features and adding new ones. This promotes code reuse and models hierarchical relationships.### PolymorphismMeans you can treat different kinds of objects in a similar way, as long as they share a common interface. This provides flexibility and adaptability in your code.

### Why Classes Matter in Practice**Code organization** — Classes create a clear structure for your code. Instead of scattering variables and functions throughout your script, you group them together within a class, making your code easier to read, understand, and maintain. Imagine a game with multiple characters, each with name, health, strength, speed — without classes you'd end up with a mess of variables like `character1_name`, `character2_health`. With a `Character` class, you neatly encapsulate all these attributes within a single object.**Reusability** — Once you've defined a class, you can create multiple instances (objects) of it without rewriting the same code repeatedly. A `BankAccount` class lets you create objects for different accounts (`my_account`, `your_account`), each with its own balance and transaction history — the deposit/withdraw logic is encapsulated once.**Encapsulation (in practice)** — Python achieves this mostly through convention rather than strict enforcement: a single underscore (`_`) prefix on an attribute name signals it's intended for internal use within the class. Python doesn't prevent direct access to these "private" attributes from outside, but the underscore is a clear signal to other developers that they shouldn't be modified directly.
