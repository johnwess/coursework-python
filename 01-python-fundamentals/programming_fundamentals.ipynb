## Conditional Statements

Conditional statements are the "if-then-else" logic of programming. They empower your code to assess a situation and select a course of action based on whether a condition is true or false. For instance, a weather app could use conditionals to advise "Wear a jacket!" if the temperature drops below a certain point, or suggest lighter attire otherwise.

In Python, the cornerstone of conditional statements is the if statement.

```python
while True:
    try:
        age = int(input("Enter your age: "))
        if age <= 0:
            print("Invalid age. Age must be a positive integer.")
            continue
        break
    except ValueError:
        print("Invalid input. Please enter a valid age.")
if age < 12:    
    price = 8
elif age < 65:
    price = 12
else:
    price = 10
print(f"The ticket price is: ${price}")
```

#### Create a function that returns an individual's name formatted as a greeting

```python
def greet(name):
    """ A greeting message including a person's name """

    # Combine the word "Hello" with the person's name and an exclamation point to create the greeting message
    name = "Hello " + str(name) + "!"

    # Return the greeting message so it can be used elsewhere in the code
    return name
```

##### Call the greet function with the name "Alice" and store the returned greeting message in the variable `result`


```python
# Call the greet function, store result in result
result = greet("Alice")

# Display result
print(result)
```

#### Create a function with a function nested inside

```python
def outer_function(x):
    """
    Outer function that takes an integer x, adds 5 to it, and then calls an inner function.

    Args:
        x (int): The input value that will be used in calculations.
    """
    y = x + 5  # Add 5 to x and store it in y

    def inner_function():
        """
        Inner function that multiplies y by 2 and prints the result.
        """
        z = y * 2  # Multiply y by 2 and store it in z
        print("Inside inner_function, z =", z)  # Print the value of z

    inner_function()  # Call the inner_function
    print("Inside outer_function, y =", y)  # Print the value of y after calling inner_function
```

##### Call the outer function

```python
# Call the outer_function with an argument of 10
outer_function(10)
```
