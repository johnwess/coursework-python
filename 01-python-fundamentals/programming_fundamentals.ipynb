## Common code execution pitfalls: How to avoid them

### Understanding Code Execution

Before your code ever runs, Python parses and compiles the entire file into bytecode. This is why a syntax error on line 200 will stop your program before line 1 ever executes — Python already scanned the whole thing and choked on it during this compile pass. So in this narrow sense, the instructor isn't wrong: there IS a full-file pass before execution.

Once compiled to bytecode, execution does proceed sequentially, top to bottom, one statement at a time — with real, immediate effects as it goes. If line 5 throws a runtime error (not a syntax error — think dividing by zero, or calling an undefined variable), lines 1–4 already ran and did whatever they did. Line 6 never runs. That's genuinely sequential, line-by-line behavior, not "read it all, then run it all."

### 2. Variable Scope

In Python, variables have a scope, which defines the region of your code where the variable is accessible. A common mistake is trying to access a variable outside its scope.

An example in code:

````python
flag = False
if flag:
    x = 10
    print(x) # No problem here

print(x) # No error if flag = True
````

**john: So variables out of scope errors are caught during compilation?**

_Claude: Thought for 6s_

Good follow-up — and the answer is a nuanced "no," which trips people up because it seems like it should be a compile-time check.

Short answer: scope is determined at compile time, but the "does this variable actually exist" check happens at runtime.

Here's the split:

Compile time: Python's compiler looks at your function and decides, for each name, whether it's local or global/nonlocal — based on whether you assign to it anywhere in that function. This is static analysis, done before anything runs.
Runtime: Whether that variable actually has a value when a given line executes — that's only checked when execution reaches that line.

This produces a classic gotcha:

```python
def foo():
    print(x)   # NameError/UnboundLocalError happens HERE, at runtime
    x = 5
```

The compiler sees x = 5 somewhere in foo(), so it decides x is a local variable for the entire function (that decision is compile-time). But when execution actually hits print(x), x hasn't been assigned yet — so Python throws UnboundLocalError at that moment, not before the function even runs.

Notice: this code would compile just fine and the interpreter would happily start executing foo(). It only blows up once execution reaches the specific line that references the not-yet-bound variable — which is a runtime event, tied to the sequential, top-to-bottom execution we just talked about.

Contrast that with a true compile-time error:

```python
def foo(:
    print("hi")
```
That's a SyntaxError — malformed code structure — and it's caught during the compile pass, before a single line executes, exactly like your instructor's quote described.

So the rule of thumb: syntax problems (mismatched parens, bad indentation, invalid tokens) → compile time, whole file, before anything runs. Name/scope/value problems (using an undefined variable, dividing by zero, calling a method that doesn't exist) → runtime, sequential, only when that line is actually reached.

**gotcha...does python allow a compile check without running code?**

_Yes — a few ways, depending on how deep a check you want:_

1. py_compile — compiles without executing

```bash
python3 -m py_compile your_script.py
```
This runs the full compile pass (syntax check + turns it into bytecode) and reports errors, but never executes a single line. It'll catch things like mismatched parentheses, bad indentation, invalid syntax — exactly the "whole file, before anything runs" category we talked about. It'll silently succeed and produce a .pyc file in a __pycache__ folder if everything's clean.

2. The compile() built-in, from within Python itself

```python
with open("your_script.py") as f:
    source = f.read()
compile(source, "your_script.py", "exec")
```
Same idea — this raises a SyntaxError if something's malformed, without executing anything. Useful if you want to check code programmatically rather than from the command line.

3. ``python3 -m py_compile`` vs. just running it — the key difference
Running python3 your_script.py normally also compiles first, but then immediately executes too. The -m py_compile route stops right after compilation — it's the cleanest way to isolate "does this even parse" from "does this actually work."

One important caveat, tying back to our UnboundLocalError example:
None of these compile-only checks will catch scope/name errors, undefined variables, type mismatches, or anything that only manifests when code actually runs. A file can compile perfectly clean and still blow up the moment you execute it — because, as we established, that class of error is a runtime event, not a compile-time one.

This is also part of why linters (like pylint or flake8) exist — they go a step further than the compiler by doing static analysis to guess at likely runtime problems (like "hey, you're using x before you define it") without actually running your code either. They're not the compiler, but they're trying to catch the gap between "compiles fine" and "actually works."

**Soooo...if the error shows a Traceback (most recent call last) in the error output, that's runtime. Compile errors don't seem to show "Traceback"**

_Exactly right — that's a genuinely reliable tell, and good pattern-matching on your part._

Why this makes sense given what we covered:

A traceback is, by definition, a call stack — it's showing you the chain of function calls that were in progress when the error hit: "this line called that function, which called this other function, which is where it blew up." That concept only makes sense if code was actually executing — you can't have a call stack for code that never ran.

SyntaxError — no traceback, because nothing was running yet:

```  File "script.py", line 3
    def foo(:
             ^
SyntaxError: invalid syntax
```
Just a file/line pointer and a caret showing where the parser choked. No "most recent call last," because there's no call history — execution never started.

Runtime error — traceback, because something was mid-execution:

```Traceback (most recent call last):
  File "script.py", line 5, in <module>
    foo()
  File "script.py", line 2, in foo
    print(x)
NameError: name 'x' is not defined
```
Here you can see the actual chain: the module called foo(), which was executing print(x) when it died. That's a live snapshot of execution in progress — which only exists because execution was in progress.

One small wrinkle worth knowing, so this heuristic doesn't fail you later: IndentationError and TabError are technically subclasses of SyntaxError (still compile-time, still no traceback) — but they can look surprisingly similar to runtime output at a glance since Python's error messages have evolved to be more descriptive. If you ever see one and it's ambiguous, checking for the literal word "Traceback" at the top is still your fastest and most reliable signal, exactly like you said.

### List Comprehensions: A Concise Approach

A list comprehension is a concise and powerful way in Python to create new lists based on existing iterables (like lists, tuples, strings, ranges, etc.). It provides a more readable and often more efficient alternative to using traditional for loops for creating lists, especially when the logic for creating the new list is relatively straightforward:

[expression for item in iterable]

Let's break down each part:

[ ... ]: The square brackets signify that the result will be a new list.

expression: This is the operation or transformation that is applied to each item in the iterable. The result of this expression for each item will become an element in the new list.

for item in iterable: This part specifies the iterable you are working with and assigns each element in the iterable to the variable item during each iteration. It's similar to the beginning of a for loop.  

Scenario: You have a list of scores for students on an examination, and you'd like to curve the grades by adding 10 points to each score. You will store this in a new list:

````python
exam_scores = [55, 70, 78, 52, 68]
curve_amount = 10
# Use a list comprehension to create a new list of curved grades
curved_grades = [score + curve_amount for score in exam_scores]
print("Original scores:", exam_scores)
print("Curved scores:", curved_grades)
````

### The DRY Principle. 

The DRY principle is a fundamental philosophy that underpins efficient coding. It advocates for eliminating redundant code, encouraging you to consolidate repetitive instructions into a single, reusable block. This approach isn't merely about saving keystrokes; it's about reducing the surface area for errors, inconsistencies, and maintenance headaches.

Let's say you're developing an e-commerce platform. Without DRY, you might find yourself copying and pasting code for tasks like calculating discounts, validating payment information, or sending order confirmation emails. This scattered repetition makes your codebase bloated and prone to errors. If a change is required, you'd have to hunt down and modify every instance of the duplicated code, increasing the risk of introducing bugs.

DRY, on the other hand, encourages you to create a centralized function for each of these tasks. For example, a ``calculate_discount(price, discount_percentage)`` function would encapsulate the logic for calculating discounts, regardless of the specific product or promotion. This function acts as a single source of truth, ensuring consistency and simplifying updates.

## Building and Using Classes for Object-Oriented Design

a `class` is a blueprint for creating objects that bundle data (attributes) and behavior (methods) together. You define a class using the class keyword followed by the class name and a colon.

### Objects

They represent real-world entities or abstract concepts within your code. Each object is like a self-contained unit with its own data and the ability to perform actions.
### Classes

They serve as blueprints or templates for creating objects. Classes define the common characteristics (attributes) and behaviors (methods) that all objects of that class share. 
### Encapsulation

This means keeping the details of how an object works hidden inside the object, so you can change those details without breaking the rest of your program. It bundles data and the actions that operate on that data within a class, protecting the internal state of an object. 
### Abstraction

This simplifies complex systems by focusing on essential features and interactions, hiding unnecessary details. It focuses on what an object does rather than how it does it, so your code is easier to understand and use.
### Inheritance

This lets you create new kinds of objects that are based on existing ones, inheriting their features and adding new ones. This promotes code reuse and models hierarchical relationships. 
### Polymorphism

This means you can treat different kinds of objects in a similar way, as long as they share a common interface. This provides flexibility and adaptability in your code. 

### Code organization

Classes create a clear structure for your code. Instead of scattering variables and functions throughout your script, you can group them together within a class, making your code easier to read, understand, and maintain.

For example, imagine you're building a game with multiple characters. Each character might have attributes like name, health, strength, and speed. Without classes, you might end up with a mess of variables like character1_name, character2_health, and so on. With a Character class, you can neatly encapsulate all these attributes within a single object.
### Reusability

Once you've defined a class, you can create multiple instances (objects) of that class. This means you don't have to rewrite the same code repeatedly. You can simply create a new object and customize it with specific attributes.

Let's say you have a BankAccount class. You can create objects for different accounts (my_account, your_account, etc.), each with its own balance and transaction history. The underlying logic for depositing and withdrawing money is encapsulated within the class, so you don't have to repeat it for each account.
### Encapsulation

Encapsulation in Python is achieved primarily through a convention rather than strict enforcement. It revolves around using a single underscore (_) prefix for attribute names to signal that they are intended for internal use within the class. While Python doesn't prevent direct access to these "private" attributes from outside the class, the underscore serves as a clear indication to other developers that these attributes should be treated as part of the class's internal implementation and not modified directly.
